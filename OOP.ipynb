```
class auto:
  def __init__(self, model, motor, barva):
    self.model = model
    self.motor = motor
    self.barva = barva
octavia = auto("octavia", "benzin", "cervena")
karoq = auto("karoq", "diesel", "modra")
fabia = auto("fabia", "benzin", "bila")
print(octavia.model)
print(karoq.motor)
print(fabia.barva)

class auto:
  def __init__(self, model):
    self, auto.model = model
  def motor(self):
    print(self.motor)
  def model(self):
    print(self.model)
  def barva(self):
    print(self.barva)





```
